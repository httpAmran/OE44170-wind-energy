# Part 3: Wind Turbine Technology [20 pts]

Rotor diameter 220 m, rated power 14.1 MWe at 11 m/s, drivetrain efficiency 97%, air density 1.205 kg/m^3.

```python
import sys
sys.path.append("../../src")  # lets `import windtools` work with plain python, no install needed

import numpy as np
from windtools import rotor_model

RHO_AIR = 1.205
RADIUS = 220 / 2
RATED_POWER_ELEC_W = 14.1e6
DRIVETRAIN_EFF = 0.97
RATED_WIND_SPEED = 11
```

## Q1: Cp from u1 = 10 m/s, u2 = 7 m/s (actuator disk theory) [5 pts]

```python
a = rotor_model.induction_factor_from_speeds(10, 7)
cp = rotor_model.cp_actuator_disk(a)
print(f"a = {a:.4f}, Cp = {cp:.4f}")
```

## Q2: U = 15 m/s, omega = 7.8 rpm -> tip speed ratio and Cp (empirical curve, beta = 0) [5 pts]

```python
lam = rotor_model.tip_speed_ratio(7.8, RADIUS, 15)
cp = rotor_model.cp_empirical(beta=0, lam=lam)
print(f"lambda = {lam:.3f}, Cp = {cp:.4f}")
```

## Q3: U = 9 m/s, omega = 5.4 rpm -> expected Cp [5 pts]

TODO: compute lambda and Cp as above, and discuss where this operating point sits relative to lambda_opt = 6.9 / Cp_max = 0.48 (e.g. below-rated region, pitch likely still at beta = 0).

```python
lam_q3 = rotor_model.tip_speed_ratio(5.4, RADIUS, 9)
cp_q3 = rotor_model.cp_empirical(beta=0, lam=lam_q3)
print(f"lambda = {lam_q3:.3f}, Cp = {cp_q3:.4f}")
```

## Q4: Thrust force at rated conditions [5 pts]

Get Cp at rated conditions from the rated aerodynamic power, convert to axial induction factor and Ct, then compute thrust.

```python
rated_power_aero_w = RATED_POWER_ELEC_W / DRIVETRAIN_EFF
cp_rated = rated_power_aero_w / (0.5 * RHO_AIR * np.pi * RADIUS**2 * RATED_WIND_SPEED**3)
a_rated = rotor_model.axial_induction_from_cp(cp_rated)
ct_rated = rotor_model.ct_from_induction(a_rated)
thrust_n = rotor_model.thrust_force(ct_rated, RHO_AIR, RADIUS, RATED_WIND_SPEED)
print(f"Cp_rated = {cp_rated:.4f}, a = {a_rated:.4f}, Ct = {ct_rated:.4f}, Thrust = {thrust_n/1e3:.1f} kN")
```
